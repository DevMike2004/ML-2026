## Question 2

Hard voting uses a regular voting system where models put their answer in and whichever answer is 
the most common in that row is used. Soft voting uses a model's confidence to get an answer. The model
with the highest confidence carries much more weight in the vote. 
Calibration matters because soft voting assumes every model's predicted prob's are on the same scale, 
which only holds if each model is well calibrated. This means, a predicted probability of .8 really
corresponds to about an 80% chance of the outcome. Heterogeneous models produce probabilities in
very different ways. Log Reg is usually fialy well calibrated, decision trees are often overconfident
and SVM's come from an extra fitted approximation. As a result a poorly calibrated, overconfident model
can dominate the averaged probabilities and overrule more accurate models, which is why
calibrating models matter before soft voting.

## Question 3

Because each tree in a bagging classifier is trained on a sample drawn with replacement, about 1/3
of the training rows are left out of any tree's samples. The oob rows act as unseen data for that tree.
The oob score is calculated by predicting each training row using only the trees that never
saw it and then measuring the accuracy of those predictions. This produces a validation estimate
from the training data alone. Making it an efficient alternative to cross-validation. No data
has to be help back and no models have to be retrained across multiple folds.
