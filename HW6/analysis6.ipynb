## Question 3

Fitting PCA on the full dataset lets the test images help decide which components are kept and how the data is centered.
The test set then influences the features the classifier is trained on so it's no longer truly "unseen".
This is data leakage. The usual result is a test accuracy that is overly optimistic and doesn't relect how
the model would do on new faces it has ever seen. In the assignment, we avoid this by putting the model inside a pipeline.

## Question 5

My results actually showed PCA and classical classifiers did pretty well with this data, having 97.5% accuracy.
The real issue is when you scale up the data. In more complex computer vision tasks, the performance would not carry over.
One problem here is maing the classifier more complex, causes features to bottleneck and the model overfits.
The results show Olivetti is just a simple dataset. It's too easy for the model for the same reasons discussed above.
PCA's kind of compression is also too universal for facial recognition. Every picture contains the same size of pixels.
This could cause the model to not actually understand what an eye looks like, but it memorizes where the thing 
it knows as an eye is seen in the pictures provided. If you gave it a large image resolution with a different facial alignment, 
it would most likely fail the recognize it.
